Installing Libraries

```python
import pandas as pd
import altair as alt
from vega_datasets import data as vega_data
```

Load the Seattle Circulation data from Github

```python
import requests
import io

url = "https://raw.githubusercontent.com/manika-lamba/SP26-LIS4_5693/refs/heads/main/lab-assignments/lab-3/Seattle-Library_2015-2021.csv"
response = requests.get(url)
response.raise_for_status() # Raise an exception for HTTP errors
text = response.text
```

Save it as a dataframe using variable df

```python
seattle_df = pd.read_csv(io.StringIO(text))
seattle_df.head()
```

I defined the chart with 'pointchart' so I could concatenate it later. I chose to use a scatterplot to compare publication year of books with number of checkouts. I followed the lab example to make the chart have a tooltip and be interactive. I defined PublicationYear with :0 for ordinal (I tried temporal at first but it looked really weird/off), and checkouts with :Q for quantitative.

```python
pointchart = alt.Chart(seattle_df).mark_point().encode(
                                    x= 'PublicationYear:O',
                                    y= 'Checkouts:Q',
                                    tooltip= ['Title', 'PublicationYear', 'Checkouts']
                                    ).interactive()

pointchart
```

```python
linechart = alt.Chart(seattle_df).mark_line().encode(
    x= 'PublicationYear:O',
    y= 'Checkouts:Q',
    tooltip= ['Title', 'PublicationYear', 'Checkouts']
).interactive()

linechart
```

```python
alt.hconcat(pointchart, linechart)
```
